### 1 . Create a Series named 'daily_sales_inr' in a notebook named 'kirana_pandas.ipynb' holding 7 values for Monday to Sunday of a fictional kirana store (your own numbers between 8,000 and 25,000), with the weekday names as the index. Print its `index`, `values` and `dtype`, then print only Wednesday's value using its label 

```python
import pandas as pd

daily_sales_inr = pd.Series(
    [12500, 14800, 17300, 11200, 19600, 22500, 15800],
    index=["Monday", "Tuesday", "Wednesday",
           "Thursday", "Friday", "Saturday", "Sunday"],
    name="daily_sales_inr"
)

print("Index:", daily_sales_inr.index)
print("Values:", daily_sales_inr.values)
print("Data type:", daily_sales_inr.dtype)

print("Wednesday sales:", daily_sales_inr["Wednesday"])
```

### 2 . Build a DataFrame named 'products_df' from a dictionary in the same notebook, with 8 products you invent and the columns product_id, product_name, category, price_inr and stock_qty (at least 3 categories). Print its `shape` and `dtypes`, then pull out `price_inr` and print `type()` of it to confirm it is a Series.

```python
products_data = {
    "product_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "product_name": ["Milk", "Bread", "Rice", "Juice",
                     "Tea", "Soap", "Shampoo", "Sugar"],
    "category": ["Food", "Food", "Food", "Drinks",
                 "Drinks", "Personal Care", "Personal Care", "Food"],
    "price_inr": [30, 25, 60, 40, 100, 35, 80, 45],
    "Quantity": [20, 15, 50, 25, 10, 30, 12, 40]
}

products_df = pd.DataFrame(products_data)

print("Shape:", products_df.shape)
print(products_df.dtypes)

price = products_df["price_inr"]
print("Type:", type(price))
```

### Create 'week_df' by placing 'daily_sales_inr' side by side with two new Series: customers (random integers from 60 to 200 with `np.random.seed(42)`) and avg_bill_inr (sales divided by customers, rounded to 0 decimals). Print 'week_df' and its `shape`, which must be (7, 3). Hint: `pd.DataFrame({'sales_inr': s1, 'customers': s2})` lines Series up by their index, so give the new Series the same index. 

```python

import numpy as np

np.random.seed(42)

customers = pd.Series(
    np.random.randint(60, 201, size=7),
    index=daily_sales_inr.index
)
 
avg_bill_inr = (daily_sales_inr / customers)                                                       # Calculate average bill


week_df = pd.DataFrame({                                                                            # create dataframe 
    "sales_inr": daily_sales_inr,
    "customers": customers,
    "avg_bill_inr": avg_bill_inr
})

print(week_df)
print("Shape:", week_df.shape)
```
