### 3. Create 'big_bills' with boolean conditions: bills paid by UPI with qty of 5 or more and unit_price_inr above 200, or bills paid by Card with qty of 8 or more. Print its row count. Hint: put every condition in its own parentheses and join them with `&` and `|`. Constraint: Do not use loops

```python
big_bills = sales_df[
    ((sales_df['payment_mode'] == 'UPI') &
     (sales_df['qty'] >= 5) &
     (sales_df['unit_price_inr'] > 200)) |
    ((sales_df['payment_mode'] == 'Card') &
     (sales_df['qty'] >= 8))
]

print(big_bills)
print("Row count:", big_bills.shape[0])
```

### 4.Rebuild the same rows as 'big_query' using `query()` with the thresholds stored in Python variables and referred to with `@`. Print whether `big_bills.equals(big_query)` is True, then save 'big_query' as 'kirana_bigbills.csv' with `index=False`. Run Restart & Run All before submitting.

```python
upi_qty = 5
upi_price = 200
card_qty = 8

big_query = sales_df.query(
    "(payment_mode == 'UPI' and qty >= @upi_qty and unit_price_inr > @upi_price) or "
    "(payment_mode == 'Card' and qty >= @card_qty)"
)

print(big_bills.equals(big_query))

big_query.to_csv('kirana_bigbills.csv', index=False)
```
