### 1 . Generate 120 members for a fictional gym in a notebook named 'gym_members.ipynb' with `np.random.seed(42)`: member_id 1 to 120, plan (Silver, Gold or Platinum), monthly_fee_inr (800, 1200 or 1800 matching the plan), months_paid 1 to 12, age 18 to 60 and trainer_sessions 0 to 20. Save it as 'gym_members_raw.csv' with `index=False`. Hint: pick plans with `np.random.choice`, then set the fee with nested `np.where`

```python
import numpy as np
import pandas as pd

np.random.seed(42)

gym_members = pd.DataFrame({
    'member_id': np.arange(1, 121),
    'plan': np.random.choice(['Silver', 'Gold', 'Platinum'], 120),
    'months_paid': np.random.randint(1, 13, 120),
    'age': np.random.randint(18, 61, 120),
    'trainer_sessions': np.random.randint(0, 21, 120)
})

gym_members['monthly_fee_inr'] = np.where(
    gym_members['plan'] == 'Silver', 800,
    np.where(gym_members['plan'] == 'Gold', 1200, 1800)
)

gym_members.to_csv('gym_members_raw.csv', index=False)

print(gym_members.head())
print("Dataset shape:", gym_members.shape)
print("CSV file saved successfully!")
```

### 2. Add two new columns to the DataFrame: 'total_paid_inr' as monthly_fee_inr times months_paid, and 'gst_inr' as 18% of total_paid_inr rounded to 2 decimals. Print the first 5 rows and the sum of total_paid_inr.

```python
gym_members['total_paid_inr'] = gym_members['monthly_fee_inr'] * gym_members['months_paid']

gym_members['gst_inr'] = (gym_members['total_paid_inr'] * 0.18).round(2)

print(gym_members.head(5))

print("Total Paid Amount:", gym_members['total_paid_inr'].sum())
```

### 3. Create 'age_group' with `apply` and a `lambda` on the age column: Youth below 25, Adult from 25 to 40 and Senior above 40. Print the number of members in each group with `value_counts()`. Hint: a lambda can hold a nested conditional such as `lambda a: 'x' if a < 1 else ('y' if a < 2 else 'z')`

```python
gym_members['age_group'] = gym_members['age'].apply(
    lambda a: 'Youth' if a < 25 else ('Adult' if a <= 40 else 'Senior')
)

print(gym_members['age_group'].value_counts())
```

### 4.Build 'gym_display' by applying a formatting function with `applymap` to monthly_fee_inr, total_paid_inr and gst_inr so each number becomes text like 'Rs 1,200'. Save the numeric DataFrame (not 'gym_display') as 'gym_members_v1.csv' with `index=False`. Hint: newer pandas versions warn that `applymap` is renamed `DataFrame.map`; use whichever runs on your install without a warning.

```python
def format_rs(x):
    return f"Rs {x:,.0f}"

gym_display = gym_members[['monthly_fee_inr', 'total_paid_inr', 'gst_inr']].map(format_rs)

print(gym_display.head())

gym_members.to_csv('gym_members_v1.csv', index=False)

print("CSV file saved successfully!")
```
