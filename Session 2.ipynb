""" 1. Generate 300 bills for a fictional kirana store in a notebook named 'kirana_io.ipynb' with `np.random.seed(42)`: bill_id 1001 to 1300, bill_date as DD/MM/YYYY text between 01/01/2026 and 28/03/2026, product from 8 names you choose, qty 1 to 10, unit_price_inr 10 to 500 and payment_mode (Cash, UPI or Card). Save it as 'kirana_sales.csv' with `index=False`. Hint: build each date with `f"{d:02d}/{m:02d}/2026"` using a random day d (1 to 28) and month m (1 to 3)."""

```python
import numpy as np
import pandas as pd

np.random.seed(42)

products = ["Rice","Wheat","Sugar","Milk","Cooking Oil", "Tea","Biscuits","Soap"]

data = []

for i in range(300):
    d = np.random.randint(1, 29)
    m = np.random.randint(1, 4)

    bill_date = f"{d:02d}/{m:02d}/2026"

    data.append({
        "bill_id": 1001 + i,
        "bill_date": bill_date,
        "product": np.random.choice(products),
        "qty": np.random.randint(1, 11),
        "unit_price_inr": np.random.randint(10, 501),
        "payment_mode": np.random.choice(["Cash", "UPI", "Card"])
    })

sales_df = pd.DataFrame(data)

sales_df.to_csv("kirana_sales.csv", index=False)

print("CSV file created successfully!")
print(sales_df.head())
```

""" 2.Load 'kirana_sales.csv' into 'sales_df' with `read_csv` and run a first inspection by printing `head(5)`, `tail(3)`, `info()` and `describe()`. Confirm in code that `sales_df.shape` is (300, 6) and that the dtype of bill_date is object (text), because dates are not converted automatically. """

```python
sales_df = pd.read_csv("kirana_sales.csv")

print(sales_df.head(5))

print(sales_df.tail(3))

print("Dataset Information:")                                                   # Dataset information
sales_df.info()

print("Statistical Description:")                                               # Statistical description
print(sales_df.describe())

print("Dataset Shape:", sales_df.shape)                                          # Confirm dataset shape
print("Shape is (300, 6):", sales_df.shape == (300, 6))

print("bill_date datatype:", sales_df["bill_date"].dtype)                        # Confirm bill_date datatype
print("bill_date is object:", sales_df["bill_date"].dtype == "object")
```
