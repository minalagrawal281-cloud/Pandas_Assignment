### 1 Load 'kirana_sales.csv' into 'sales_df' in a notebook named 'kirana_sort.ipynb'. Create 'pricey_top10' holding the 10 bills with the highest unit_price_inr, using `sort_values` in descending order, and print only the product and unit_price_inr columns of it. Hint: `ascending=False`, then `head(10)`.

```python
import pandas as pd

sales_df = pd.read_csv('kirana_sales.csv')

pricey_top10 = sales_df.sort_values(
    by='unit_price_inr',
    ascending=False
).head(10)

print(pricey_top10[['product', 'unit_price_inr']])
```

### 2 Sort 'sales_df' by product from A to Z and, within each product, by qty from highest to lowest, using one `sort_values` call with a list for `by` and a list for `ascending`. Apply `reset_index(drop=True)`, save the result as 'sales_sorted' and print `sales_sorted.head(10)` to confirm the index starts from 0 

```python
sales_sorted = sales_df.sort_values(
    by=['product', 'qty'],
    ascending=[True, False]
).reset_index(drop=True)

print(sales_sorted.head(10))
```

### 3. Create 'sales_gap' by sorting 'sales_df' with `sort_index(ascending=False)` and then calling `reindex(range(290, 310))` on the result. Print `sales_gap.isnull().all(axis=1).sum()`, which must be 10 because the labels 300 to 309 do not exist in the data.

```python
sales_gap = sales_df.sort_index(ascending=False).reindex(range(290, 310))

print(sales_gap.isnull().all(axis=1).sum())
```

### 4.Reorder the columns of 'sales_df' as bill_id, product, qty, unit_price_inr, payment_mode, bill_date using `reindex(columns=[...])` into 'sales_v2'. Print `list(sales_v2.columns)` and save it as 'kirana_sales_v2.csv' with `index=False`. Run Restart & Run All before submitting

```python
sales_v2 = sales_df.reindex(columns=['bill_id','product','qty', 'unit_price_inr','payment_mode','bill_date'])

print(list(sales_v2.columns))

sales_v2.to_csv('kirana_sales_v2.csv', index=False)
```
