### 1.Load 'gym_members_v2.csv' into 'gym_df' in a notebook named 'gym_dates.ipynb', add 'join_date' as DD/MM/YYYY text (random day 1 to 28, month 1 to 12, year 2024 or 2025, `np.random.seed(42)`) and convert it with `pd.to_datetime`. Print `gym_df.dtypes` to show datetime64 for join_date. Hint: pass `dayfirst=True` or `format='%d/%m/%Y'`, otherwise 05/03/2025 is read as 3 May

```python
import numpy as np
import pandas as pd

gym_df = pd.read_csv('gym_members_v2.csv')

np.random.seed(42)

day = np.random.randint(1, 29, 120)
month = np.random.randint(1, 13, 120)
year = np.random.choice([2024, 2025], 120)

gym_df['join_date'] = (
    pd.Series(day).astype(str).str.zfill(2) + '/' +
    pd.Series(month).astype(str).str.zfill(2) + '/' +
    pd.Series(year).astype(str)
)

gym_df['join_date'] = pd.to_datetime(
    gym_df['join_date'],
    format='%d/%m/%Y'
)

print(gym_df.head())

print(gym_df.dtypes)
```

### 2.Extract 'join_year', 'join_month' and 'join_weekday' (the day name, such as Monday) from join_date with the `dt` accessor, then print the number of members who joined in each year using `value_counts()`. Hint: `dt.day_name()` returns the weekday name.

```python
gym_df['join_year'] = gym_df['join_date'].dt.year                                                      # Extract year, month and weekday

gym_df['join_month'] = gym_df['join_date'].dt.month

gym_df['join_weekday'] = gym_df['join_date'].dt.day_name()

print(gym_df['join_year'].value_counts())
print(gym_df.head())
```

### 3.Create 'month_starts' with `pd.date_range` for 2025 (12 values, `freq='MS'`), then build 'h1_joiners' with the members whose join_date is on or after month_starts[0] and before month_starts[6]. Print the row count of 'h1_joiners'.

```python
month_starts = pd.date_range(start='2025-01-01', periods=12, freq='MS')                               # Create month starts for 2025

h1_joiners = gym_df[                                                                                        # Select members who joined in the first 6 months of 2025
    (gym_df['join_date'] >= month_starts[0]) &
    (gym_df['join_date'] < month_starts[6])
]

print("H1 Joiners:", len(h1_joiners))
```

### 4.Calculate tenure: set `as_of = pd.Timestamp('2026-01-01')`, add 'tenure_days' (the date difference with `dt.days`) and 'tenure_months' (tenure_days divided by 30 using `//`), print the minimum and maximum tenure_days, and save the DataFrame as 'gym_members_v3.csv' with `index=False`. Constraint: Do not loop over rows.

```python
as_of = pd.Timestamp('2026-01-01')

gym_df['tenure_days'] = (as_of - gym_df['join_date']).dt.days                                          # Calculate tenure in days

gym_df['tenure_months'] = gym_df['tenure_days'] // 30                                           # Calculate tenure in months

print("Minimum tenure days:", gym_df['tenure_days'].min())
print("Maximum tenure days:", gym_df['tenure_days'].max())

gym_df.to_csv('gym_members_v3.csv', index=False)

print("CSV file saved successfully!")
```
