### 1.Generate 150 students for a fictional coaching class in a notebook named 'coaching_missing.ipynb' with `np.random.seed(42)`: student_id 1 to 150, batch (A, B or C), maths, science and english marks 35 to 100, attendance_pct 50 to 100 and fees_paid_inr 15000 to 40000. Blank 12% of maths, 8% of science and 10% of attendance_pct at random, write -1 into fees_paid_inr for 10 rows, and save 'coaching_marks_raw.csv'. Hint: `df.loc[df.sample(frac=0.12, random_state=42).index, 'maths'] = np.nan`.

```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Generate 150 students
df = pd.DataFrame({
    'student_id': range(1, 151),
    'batch': np.random.choice(['A', 'B', 'C'], 150),
    'maths': np.random.randint(35, 101, 150),
    'science': np.random.randint(35, 101, 150),
    'english': np.random.randint(35, 101, 150),
    'attendance_pct': np.random.randint(50, 101, 150),
    'fees_paid_inr': np.random.randint(15000, 40001, 150)
})

# Add missing values (NaN)
df.loc[df.sample(frac=0.12, random_state=42).index, 'maths'] = np.nan

df.loc[df.sample(frac=0.08, random_state=42).index, 'science'] = np.nan

df.loc[df.sample(frac=0.10, random_state=42).index, 'attendance_pct'] = np.nan

# Add -1 in fees for 10 random rows
fees_index = df.sample(n=10, random_state=10).index
df.loc[fees_index, 'fees_paid_inr'] = -1

# Save CSV
df.to_csv('coaching_marks_raw.csv', index=False)

# Display data
print(df.head(10))
print(df.shape)
```

### 2.Load 'coaching_marks_raw.csv', turn the -1 placeholders into NaN with `replace(-1, np.nan)`, and build 'missing_report' with the columns missing_count and missing_pct (1 decimal) for every column using `isnull()`. Also print how many rows have at least one missing value. Hint: `isnull().any(axis=1).sum()` counts those rows.

```python

df = pd.read_csv('coaching_marks_raw.csv')                                                  # Load CSV

df = df.replace(-1, np.nan)                                                                          # Replace -1 with NaN


missing_report = pd.DataFrame({
    'missing_count': df.isnull().sum(),                                                         # Create missing report
    'missing_pct': (df.isnull().mean() * 100).round(1)
})

print(missing_report)
                                                                                                        # Count rows having at least one missing value
print("Rows with missing values:", df.isnull().any(axis=1).sum())
```

### 3.Create 'marks_filled' by filling maths with its median, science with its mean rounded to 1 decimal, attendance_pct with its median and fees_paid_inr with its median, using `fillna`. Print `isnull().sum().sum()`, which must be 0. Constraint: Assign each column back with `df['col'] = df['col'].fillna(...)`; do not use `inplace=True` on a column.

```python
marks_filled = df.copy()

marks_filled['maths'] = marks_filled['maths'].fillna(marks_filled['maths'].median())

marks_filled['science'] = marks_filled['science'].fillna(round(marks_filled['science'].mean(), 1))

marks_filled['attendance_pct'] = marks_filled['attendance_pct'].fillna(marks_filled['attendance_pct'].median())

marks_filled['fees_paid_inr'] = marks_filled['fees_paid_inr'].fillna(marks_filled['fees_paid_inr'].median())

print(marks_filled.isnull().sum().sum())
```

### 4.Create 'marks_dropped' with `dropna(subset=['maths', 'science'])`, then print a 3-row table 'compare_df' with the version name (raw, dropped, filled), row count and mean maths for each. Finish by saving 'marks_filled' as 'coaching_marks_clean.csv' with `index=False`. Run Restart & Run All before submitting.

```python
marks_dropped = df.dropna(subset=['maths', 'science'])                                # Drop rows with missing maths or science

compare_df = pd.DataFrame({                                                         # comparison DataFrame
    'version': ['raw', 'dropped', 'filled'],
    'row_count': [len(df), len(marks_dropped),len(marks_filled)],
    'mean_maths': [ df['maths'].mean(), marks_dropped['maths'].mean(), marks_filled['maths'].mean()]
})

print(compare_df)

marks_filled.to_csv('coaching_marks_clean.csv', index=False)
```
