### 1 . Load 'gym_members_v1.csv' into 'gym_df' in a notebook named 'gym_text.ipynb' and add two messy text columns with `np.random.seed(42)` and `np.random.choice`: member_name from 10 invented two-word names typed in mixed styles such as 'riya SHAH', ' KARAN patel ' and 'Neha Joshi', and fitness_goal from 'Weight Loss', 'weight-loss', 'WEIGHT LOSS', 'Muscle Gain', 'muscle_gain' and 'Stamina'. Save it as 'gym_members_messy.csv'. Hint: if you missed the last session, first create 120 rows with member_id and plan.

```python
import numpy as np
import pandas as pd

gym_df = pd.read_csv('gym_members_v1.csv')

np.random.seed(42)

names = ['riya SHAH',' KARAN patel ','Neha Joshi','rahul MEHTA','PRIYA desai','Amit Kumar',' pooja shah ','VIKAS Patel','simran JOSHI','Raj Mehta']

gym_df['member_name'] = np.random.choice(names, 120)                                # Add messy text columns

gym_df['fitness_goal'] = np.random.choice(
    ['Weight Loss', 'weight-loss', 'WEIGHT LOSS',
     'Muscle Gain', 'muscle_gain', 'Stamina'],
    120
)

gym_df.to_csv('gym_members_messy.csv', index=False)

print(gym_df.head())

print("Dataset shape:", gym_df.shape)
print("CSV file saved successfully!")
```

### 2. Create 'name_clean' from member_name using `str.strip()`, a regex `str.replace` that collapses repeated spaces, and `str.title()`. Print the unique values of 'name_clean'; there must be no more than 10. Hint: `str.replace(r'\s+', ' ', regex=True)` collapses runs of spaces.

```python
gym_df['name_clean'] = (
    gym_df['member_name']
    .str.strip()
    .str.replace(r'\s+', ' ', regex=True)
    .str.title()
)

print(gym_df['name_clean'].unique())

print("Total unique names:", gym_df['name_clean'].nunique())
```

### 3.Standardise fitness_goal into 'goal_clean' with `str.lower()` and `str.replace` so that '-' and '_' become spaces, then use `str.contains('weight')` on it to print how many members want weight loss. Check that `goal_clean.value_counts()` shows exactly 3 goals

```python
gym_df['goal_clean'] = (
    gym_df['fitness_goal']
    .str.lower()
    .str.replace('-', ' ', regex=False)
    .str.replace('_', ' ', regex=False)
)

print(gym_df['goal_clean'].value_counts())

weight_loss_count = gym_df['goal_clean'].str.contains('weight').sum()

print("Members who want Weight Loss:", weight_loss_count)
```

### 4. Split 'name_clean' into 'first_name' and 'last_name' with `str.split(' ', expand=True)`, drop the raw member_name and fitness_goal columns, and save the DataFrame as 'gym_members_v2.csv' with `index=False`. Print its first 5 rows. Hint: `df[['first_name', 'last_name']] = df['name_clean'].str.split(' ', expand=True)`.

```python

gym_df[['first_name', 'last_name']] = gym_df['name_clean'].str.split(' ', expand=True)
                                                                                                   # Split name into first and last name

gym_df.drop(columns=['member_name', 'fitness_goal'], inplace=True)                                 # Drop raw text columns

gym_df.to_csv('gym_members_v2.csv', index=False)

print(gym_df.head(5))
```
